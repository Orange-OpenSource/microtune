```python
print("Dataframe loaded with shape:", df11.shape)
buf_sizes_list = df11["buf_size"].unique().tolist()
print("Count Buffer sizes found:", len(buf_sizes_list))
list = df11["origin"].unique().tolist()
print("Origins found:", list)
list = df11["combined_column"].unique().tolist()
print("v11 combined_column found:", list)
list = df11["combined_column"].unique().tolist()
print("v11 combined_column found:", list)

df11 = df11[df11["perf_target_level"].isin([0.98])]
objgap = df11["objective_gap"].unique().tolist()
print(f"DF11 objective_gap: {objgap}")
print()

#df12 = obsdf12.fix
list = df12["combined_column"].unique().tolist()
print("v12 combined_column found:", list)
print(df12.columns.tolist())

objgap = df12["objective_gap"].unique().tolist()
print(f"DF12 objective_gap: {objgap}")
print()

df11["origin"] = 11
df12["origin"] = 12

df15 =  pd.concat([df11, df12], ignore_index=True)

list = df15["combined_column"].unique().tolist()
print("v15 combined_column:", list)
print("v15 combined_column count:", len(list))
print(df15.columns.tolist())

objgap = df15["objective_gap"].unique().tolist()
print(f"DF15 objective_gap: {objgap}")
print()

df15 = df15[df15["perf_target_level"].isin([0.98])]

print("Dataframe for DBSAS simulations purpose.")
obsdf15 = ObsSamplesDF(version="15")
obsdf15.saveFullPickle("../import/workloads", df15)

```
